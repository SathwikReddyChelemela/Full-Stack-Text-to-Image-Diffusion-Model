```python
# Complete list of imports and installations
!pip install -q torch torchvision transformers gradio tqdm matplotlib pillow datasets

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Dataset
from torchvision import transforms
from PIL import Image
import numpy as np
from tqdm import tqdm
import gradio as gr
from transformers import CLIPTokenizer
import matplotlib.pyplot as plt
from datasets import load_dataset
import pandas as pd
import os

# Set device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Load the Smithsonian Butterflies dataset directly from HuggingFace
print("Loading Smithsonian Butterflies dataset...")
dataset = load_dataset("huggan/smithsonian_butterflies_subset", split="train")
print(f"Dataset loaded with {len(dataset)} images")

# Load the CLIP tokenizer
tokenizer = CLIPTokenizer.from_pretrained("openai/clip-vit-base-patch32")

# Image transform to 256x256 and [-1, 1] normalization
image_transform = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.CenterCrop(256),
    transforms.ToTensor(),
    transforms.Normalize([0.5, 0.5, 0.5], [0.5, 0.5, 0.5])
])

# Define ConvVAE
class ConvVAE(nn.Module):
    def __init__(self, in_channels=3, latent_channels=4, base_channels=64):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Conv2d(in_channels, base_channels, 4, 2, 1),
            nn.GroupNorm(8, base_channels),
            nn.SiLU(),

            nn.Conv2d(base_channels, base_channels * 2, 4, 2, 1),
            nn.GroupNorm(8, base_channels * 2),
            nn.SiLU(),

            nn.Conv2d(base_channels * 2, base_channels * 4, 4, 2, 1),
            nn.GroupNorm(8, base_channels * 4),
            nn.SiLU(),

            nn.Conv2d(base_channels * 4, base_channels * 4, 4, 2, 1),
            nn.GroupNorm(8, base_channels * 4),
            nn.SiLU(),
        )
        self.to_mu = nn.Conv2d(base_channels * 4, latent_channels, 3, 1, 1)
        self.to_logvar = nn.Conv2d(base_channels * 4, latent_channels, 3, 1, 1)

        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(latent_channels, base_channels * 4, 4, 2, 1),
            nn.GroupNorm(8, base_channels * 4),
            nn.SiLU(),

            nn.ConvTranspose2d(base_channels * 4, base_channels * 2, 4, 2, 1),
            nn.GroupNorm(8, base_channels * 2),
            nn.SiLU(),

            nn.ConvTranspose2d(base_channels * 2, base_channels, 4, 2, 1),
            nn.GroupNorm(8, base_channels),
            nn.SiLU(),

            nn.ConvTranspose2d(base_channels, base_channels, 4, 2, 1),
            nn.GroupNorm(8, base_channels),
            nn.SiLU(),

            nn.Conv2d(base_channels, in_channels, 3, 1, 1),
            nn.Tanh()
        )

    def encode(self, x):
        h = self.encoder(x)
        mu = self.to_mu(h)
        logvar = self.to_logvar(h)
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        z = mu + eps * std
        return z, mu, logvar

    def decode(self, z):
        return self.decoder(z)

    def forward(self, x):
        z, mu, logvar = self.encode(x)
        recon = self.decode(z)
        return recon, mu, logvar

# Define ResidualBlock for UNet
class ResidualBlock(nn.Module):
    def __init__(self, in_channels, out_channels, time_emb_dim=512, cross_attn_dim=768):
        super().__init__()
        self.norm1 = nn.GroupNorm(8, in_channels)
        self.conv1 = nn.Conv2d(in_channels, out_channels, kernel_size=3, padding=1)

        self.norm2 = nn.GroupNorm(8, out_channels)
        self.conv2 = nn.Conv2d(out_channels, out_channels, kernel_size=3, padding=1)

        self.time_emb_proj = nn.Linear(time_emb_dim, out_channels)
        self.shortcut = nn.Conv2d(in_channels, out_channels, 1) if in_channels != out_channels else nn.Identity()

        self.use_cross_attn = cross_attn_dim is not None
        if self.use_cross_attn:
            # Simple cross-attention mechanism
            self.context_proj = nn.Linear(cross_attn_dim, out_channels)
            self.q_proj = nn.Linear(out_channels, out_channels)
            self.k_proj = nn.Linear(out_channels, out_channels)
            self.v_proj = nn.Linear(out_channels, out_channels)
            self.out_proj = nn.Linear(out_channels, out_channels)

    def forward(self, x, t_emb, context=None):
        h = F.silu(self.norm1(x))
        h = self.conv1(h)
        h = h + self.time_emb_proj(t_emb).unsqueeze(-1).unsqueeze(-1)
        h = self.conv2(F.silu(self.norm2(h)))

        if self.use_cross_attn and context is not None:
            # Simplified cross-attention
            b, c, height, width = h.shape
            h_flat = h.view(b, c, -1).permute(0, 2, 1)  # [B, H*W, C]

            # Project context
            context_proj = self.context_proj(context).unsqueeze(1)  # [B, 1, C]

            # Simple attention
            q = self.q_proj(h_flat)
            k = self.k_proj(context_proj)
            v = self.v_proj(context_proj)

            # Calculate attention scores
            attn_weights = torch.bmm(q, k.transpose(1, 2)) / (c ** 0.5)
            attn_weights = F.softmax(attn_weights, dim=-1)

            # Apply attention
            attn_output = torch.bmm(attn_weights, v)
            attn_output = self.out_proj(attn_output)

            # Add to feature map
            h = h + attn_output.permute(0, 2, 1).view(b, c, height, width)

        return h + self.shortcut(x)

# Define UNet model for diffusion with fixed tensor dimensions
class UNet(nn.Module):
    def __init__(self, latent_channels=4, base_channels=64, time_emb_dim=128, text_emb_dim=768):
        super().__init__()
        self.time_mlp = nn.Sequential(
            nn.Linear(1, time_emb_dim),
            nn.SiLU(),
            nn.Linear(time_emb_dim, time_emb_dim),
        )

        # Downsampling
        self.init_conv = nn.Conv2d(latent_channels, base_channels, kernel_size=3, padding=1)
        self.down1 = ResidualBlock(base_channels, base_channels * 2, time_emb_dim)
        self.down2 = ResidualBlock(base_channels * 2, base_channels * 4, time_emb_dim)

        # Middle with cross-attention
        self.mid = ResidualBlock(base_channels * 4, base_channels * 4, time_emb_dim, text_emb_dim)

        # Upsampling
        self.up1 = ResidualBlock(base_channels * 8, base_channels * 2, time_emb_dim)
        self.up2 = ResidualBlock(base_channels * 4, base_channels, time_emb_dim)

        self.final_conv = nn.Conv2d(base_channels, latent_channels, kernel_size=1)

    def forward(self, x, t, context=None):
        t = t.float().unsqueeze(-1) / 1000
        t_emb = self.time_mlp(t)

        h1 = self.init_conv(x)
        h2 = self.down1(h1, t_emb)
        h3 = self.down2(h2, t_emb)

        h_mid = self.mid(h3, t_emb, context)

        h = self.up1(torch.cat([h_mid, h3], dim=1), t_emb)
        h = self.up2(torch.cat([h, h2], dim=1), t_emb)

        return self.final_conv(h)

# Enhanced UNet model with better skip connections
class EnhancedUNet(nn.Module):
    def __init__(self, latent_channels=4, base_channels=64, time_emb_dim=128, text_emb_dim=768):
        super().__init__()
        self.latent_channels = latent_channels
        self.base_channels = base_channels

        self.time_mlp = nn.Sequential(
            nn.Linear(1, time_emb_dim),
            nn.SiLU(),
            nn.Linear(time_emb_dim, time_emb_dim),
            nn.SiLU(),
            nn.Linear(time_emb_dim, time_emb_dim),
        )

        # Downsampling
        self.init_conv = nn.Conv2d(latent_channels, base_channels, kernel_size=3, padding=1)

        # First downsampling block
        self.down1 = ResidualBlock(base_channels, base_channels, time_emb_dim)
        self.down1_downsample = nn.Conv2d(base_channels, base_channels * 2, 4, 2, 1)

        # Second downsampling block
        self.down2 = ResidualBlock(base_channels * 2, base_channels * 2, time_emb_dim)
        self.down2_downsample = nn.Conv2d(base_channels * 2, base_channels * 4, 4, 2, 1)

        # Middle with cross-attention
        self.mid = ResidualBlock(base_channels * 4, base_channels * 4, time_emb_dim, text_emb_dim)

        # Upsampling with clear tensor dimensions
        self.up1 = nn.ConvTranspose2d(base_channels * 4, base_channels * 2, 4, 2, 1)
        self.up1_block = ResidualBlock(base_channels * 4, base_channels * 2, time_emb_dim)

        self.up2 = nn.ConvTranspose2d(base_channels * 2, base_channels, 4, 2, 1)
        self.up2_block = ResidualBlock(base_channels * 2, base_channels, time_emb_dim)

        self.final_conv = nn.Sequential(
            nn.GroupNorm(8, base_channels),
            nn.SiLU(),
            nn.Conv2d(base_channels, latent_channels, kernel_size=3, padding=1)
        )

    def forward(self, x, t, context=None):
        # Time embedding
        t = t.float().unsqueeze(-1) / 1000
        t_emb = self.time_mlp(t)

        # Initial features
        x0 = self.init_conv(x)

        # Downsampling with stored features for skip connections
        x1 = self.down1(x0, t_emb)
        x1_down = self.down1_downsample(x1)

        x2 = self.down2(x1_down, t_emb)
        x2_down = self.down2_downsample(x2)

        # Middle block with cross-attention
        x_mid = self.mid(x2_down, t_emb, context)

        # Upsampling with proper skip connections
        x3 = self.up1(x_mid)  # Upsample
        x3 = torch.cat([x3, x2], dim=1)  # Skip connection
        x3 = self.up1_block(x3, t_emb)  # Process

        x4 = self.up2(x3)  # Upsample
        x4 = torch.cat([x4, x1], dim=1)  # Skip connection
        x4 = self.up2_block(x4, t_emb)  # Process

        # Final output
        return self.final_conv(x4)

# Simple text encoder
class SimpleTextEncoder(nn.Module):
    def __init__(self, vocab_size=49408, emb_dim=768, max_length=77):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, emb_dim)
        self.position_embedding = nn.Embedding(max_length, emb_dim)
        self.transformer = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(
                d_model=emb_dim,
                nhead=8,
                dim_feedforward=2048,
                batch_first=True
            ),
            num_layers=2
        )
        self.layer_norm = nn.LayerNorm(emb_dim)
        self.projection = nn.Linear(emb_dim, emb_dim)

    def forward(self, tokens, attention_mask=None):
        batch_size, seq_len = tokens.shape
        positions = torch.arange(0, seq_len, device=tokens.device).unsqueeze(0).repeat(batch_size, 1)

        token_embeddings = self.embedding(tokens)
        position_embeddings = self.position_embedding(positions)

        embeddings = token_embeddings + position_embeddings

        if attention_mask is not None:
            padding_mask = (attention_mask == 0)
        else:
            padding_mask = None

        # Pass through transformer
        transformer_output = self.transformer(embeddings, src_key_padding_mask=padding_mask)

        # Global pooling with attention mask
        if attention_mask is not None:
            # Masked average pooling
            mask_expanded = attention_mask.unsqueeze(-1).float()
            sum_embeddings = torch.sum(transformer_output * mask_expanded, 1)
            sum_mask = torch.sum(mask_expanded, 1)
            pooled_output = sum_embeddings / sum_mask
        else:
            # Simple mean pooling
            pooled_output = transformer_output.mean(dim=1)

        # Apply layer norm and projection
        pooled_output = self.layer_norm(pooled_output)
        return self.projection(pooled_output)

# Enhanced text encoder with better capacity for fine details
class EnhancedTextEncoder(nn.Module):
    def __init__(self, vocab_size=49408, emb_dim=768, max_length=77):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, emb_dim)
        self.position_embedding = nn.Embedding(max_length, emb_dim)

        # Use more attention heads and layers for better text understanding
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=emb_dim,
            nhead=12,  # More attention heads
            dim_feedforward=3072,  # Larger feedforward network
            dropout=0.1,  # Add some dropout for regularization
            batch_first=True
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=4)  # More layers

        self.layer_norm = nn.LayerNorm(emb_dim)
        self.projection = nn.Sequential(
            nn.Linear(emb_dim, emb_dim),
            nn.GELU(),
            nn.Linear(emb_dim, emb_dim)
        )

    def forward(self, tokens, attention_mask=None):
        batch_size, seq_len = tokens.shape
        positions = torch.arange(0, seq_len, device=tokens.device).unsqueeze(0).repeat(batch_size, 1)

        token_embeddings = self.embedding(tokens)
        position_embeddings = self.position_embedding(positions)

        embeddings = token_embeddings + position_embeddings

        if attention_mask is not None:
            padding_mask = (attention_mask == 0)
        else:
            padding_mask = None

        # Pass through transformer
        transformer_output = self.transformer(embeddings, src_key_padding_mask=padding_mask)

        # Global pooling with attention mask
        if attention_mask is not None:
            # Masked average pooling
            mask_expanded = attention_mask.unsqueeze(-1).float()
            sum_embeddings = torch.sum(transformer_output * mask_expanded, 1)
            sum_mask = torch.sum(mask_expanded, 1)
            pooled_output = sum_embeddings / sum_mask
        else:
            # Simple mean pooling
            pooled_output = transformer_output.mean(dim=1)

        # Apply layer norm and projection
        pooled_output = self.layer_norm(pooled_output)
        return self.projection(pooled_output)

# Custom dataset class for Smithsonian Butterflies
class ButterflyDataset(Dataset):
    def __init__(self, hf_dataset, tokenizer, transform=None):
        self.data = hf_dataset
        self.tokenizer = tokenizer
        self.transform = transform if transform is not None else image_transform

        # Predefined descriptive terms for better prompting
        self.color_terms = ["vibrant", "colorful", "bright", "iridescent", "beautiful"]
        self.wing_terms = ["delicate", "detailed", "patterned", "textured", "intricate"]

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        item = self.data[idx]

        # Get image and convert to RGB
        image = item["image"].convert("RGB")

        # Apply transforms
        if self.transform:
            image = self.transform(image)

        # Create caption - use name or a default caption
        base_caption = item["name"] if item["name"] else "a butterfly"

        # Generate a more detailed caption with randomization for variety
        import random
        color_desc = random.choice(self.color_terms)
        wing_desc = random.choice(self.wing_terms)
        caption = f"a detailed photo of {color_desc} {base_caption} butterfly with {wing_desc} wings"

        # Tokenize caption
        tokenized = self.tokenizer(
            caption,
            padding="max_length",
            max_length=77,
            truncation=True,
            return_tensors="pt"
        )

        return image, {
            "input_ids": tokenized.input_ids.squeeze(0),
            "attention_mask": tokenized.attention_mask.squeeze(0),
            "caption": caption  # Return caption for debugging
        }

# Collate function for batching
def collate_fn(batch):
    images, token_dicts = zip(*batch)
    input_ids = torch.stack([t["input_ids"] for t in token_dicts])
    attention_mask = torch.stack([t["attention_mask"] for t in token_dicts])
    captions = [t["caption"] for t in token_dicts]
    return torch.stack(images), {
        "input_ids": input_ids,
        "attention_mask": attention_mask,
        "captions": captions
    }

# Setup diffusion process with improved schedulers
def cosine_beta_schedule(timesteps, s=0.008):
    """
    Cosine schedule as proposed in https://arxiv.org/abs/2102.09672
    Better for image generation tasks
    """
    steps = timesteps + 1
    x = torch.linspace(0, timesteps, steps)
    alphas_cumprod = torch.cos(((x / timesteps) + s) / (1 + s) * torch.pi * 0.5) ** 2
    alphas_cumprod = alphas_cumprod / alphas_cumprod[0]
    betas = 1 - (alphas_cumprod[1:] / alphas_cumprod[:-1])
    return torch.clip(betas, 0.0001, 0.9999)

def linear_beta_schedule(timesteps, start=1e-4, end=0.02):
    """Original linear schedule"""
    return torch.linspace(start, end, timesteps)

def get_index_from_list(vals, t, x_shape):
    batch_size = t.shape[0]
    out = vals.gather(-1, t)
    return out.reshape(batch_size, *((1,) * (len(x_shape) - 1)))

def forward_diffusion_sample(x_0, t, device="cpu"):
    noise = torch.randn_like(x_0)
    sqrt_alphas_cumprod_t = get_index_from_list(sqrt_alphas_cumprod, t, x_0.shape)
    sqrt_one_minus_alphas_cumprod_t = get_index_from_list(sqrt_one_minus_alphas_cumprod, t, x_0.shape)

    return sqrt_alphas_cumprod_t * x_0 + sqrt_one_minus_alphas_cumprod_t * noise, noise

# Define diffusion parameters - using better cosine schedule
T = 1000
betas = cosine_beta_schedule(timesteps=T).to(device)  # Cosine schedule for better quality
alphas = 1.0 - betas
alphas_cumprod = torch.cumprod(alphas, dim=0)
alphas_cumprod_prev = F.pad(alphas_cumprod[:-1], (1, 0), value=1.0)
sqrt_alphas_cumprod = torch.sqrt(alphas_cumprod)
sqrt_one_minus_alphas_cumprod = torch.sqrt(1 - alphas_cumprod)
posterior_variance = betas * (1. - alphas_cumprod_prev) / (1. - alphas_cumprod)

# Create dataset and dataloader
train_dataset = ButterflyDataset(dataset, tokenizer)
train_dataloader = DataLoader(
    train_dataset,
    batch_size=4,
    shuffle=True,
    collate_fn=collate_fn
)

# Initialize models with improved architecture
vae = ConvVAE(latent_channels=4, base_channels=64).to(device)
text_encoder = EnhancedTextEncoder(
    vocab_size=len(tokenizer.get_vocab()),
    emb_dim=768,
    max_length=77
).to(device)
unet = EnhancedUNet(
    latent_channels=4,
    base_channels=64,
    time_emb_dim=128,
    text_emb_dim=768
).to(device)

# Optimizers with better learning rates
vae_optimizer = torch.optim.AdamW(vae.parameters(), lr=1e-4, weight_decay=0.01)
text_encoder_optimizer = torch.optim.AdamW(text_encoder.parameters(), lr=5e-5)
unet_optimizer = torch.optim.AdamW(unet.parameters(), lr=1e-5)

# Learning rate schedulers for better training
vae_scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(vae_optimizer, T_max=25)
text_scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(text_encoder_optimizer, T_max=30)
unet_scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(unet_optimizer, T_max=30)

# Training function for VAE
def train_vae(num_epochs=25):
    vae.train()
    best_loss = float('inf')

    for epoch in range(num_epochs):
        total_loss = 0

        for images, _ in tqdm(train_dataloader, desc=f"VAE Epoch {epoch+1}/{num_epochs}"):
            images = images.to(device)

            # Forward pass
            reconstructed, mu, logvar = vae(images)

            # Reconstruction loss
            recon_loss = F.mse_loss(reconstructed, images)

            # KL divergence with annealing
            kl_weight = min(1.0, (epoch) / 10)  # Gradual increase in KL weight
            kl_loss = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())
            kl_loss = kl_loss / (images.shape[0] * 4 * 16 * 16)  # Normalize

            # Total loss
            loss = recon_loss + kl_weight * 0.1 * kl_loss

            # Backpropagation
            vae_optimizer.zero_grad()
            loss.backward()
            # Gradient clipping for stability
            torch.nn.utils.clip_grad_norm_(vae.parameters(), 1.0)
            vae_optimizer.step()

            total_loss += loss.item()

        # Step the learning rate scheduler
        vae_scheduler.step()

        avg_loss = total_loss / len(train_dataloader)
        print(f"VAE Epoch {epoch+1}/{num_epochs}, Loss: {avg_loss:.4f}")

        # Save best model
        if avg_loss < best_loss:
            best_loss = avg_loss
            torch.save(vae.state_dict(), "butterfly_vae_best.pt")
            print(f"New best model saved with loss: {best_loss:.4f}")

        # Save checkpoint periodically
        if (epoch + 1) % 5 == 0 or epoch == num_epochs - 1:
            torch.save(vae.state_dict(), f"butterfly_vae_epoch_{epoch+1}.pt")

    # Save final model
    torch.save(vae.state_dict(), "butterfly_vae.pt")

    # Visualize some reconstructions
    vae.eval()
    with torch.no_grad():
        sample_images = next(iter(train_dataloader))[0][:4].to(device)
        reconstructed, _, _ = vae(sample_images)

        # Display original and reconstructed
        plt.figure(figsize=(12, 6))
        for i in range(4):
            # Original
            plt.subplot(2, 4, i+1)
            img = sample_images[i].cpu().permute(1, 2, 0).numpy()
            img = (img + 1) / 2  # Scale from [-1, 1] to [0, 1]
            plt.imshow(img)
            plt.title("Original")
            plt.axis("off")

            # Reconstructed
            plt.subplot(2, 4, i+5)
            recon = reconstructed[i].cpu().permute(1, 2, 0).numpy()
            recon = (recon + 1) / 2  # Scale from [-1, 1] to [0, 1]
            plt.imshow(recon)
            plt.title("Reconstructed")
            plt.axis("off")

        plt.tight_layout()
        plt.savefig("vae_reconstructions.png")
        plt.show()

    # Load the best model for diffusion training
    vae.load_state_dict(torch.load("butterfly_vae_best.pt"))
    print("Loaded best VAE model for diffusion training")

# Training function for diffusion model
def train_diffusion(num_epochs=30, gradient_accumulation_steps=2):
    # Set VAE to eval mode (we only train text encoder and UNet)
    vae.eval()
    text_encoder.train()
    unet.train()

    best_loss = float('inf')

    # EMA model for better stability
    ema_unet = EnhancedUNet(
        latent_channels=4,
        base_channels=64,
        time_emb_dim=128,
        text_emb_dim=768
    ).to(device)

    # Initialize EMA with current weights
    for param, ema_param in zip(unet.parameters(), ema_unet.parameters()):
        ema_param.data.copy_(param.data)
    ema_unet.eval()

    # EMA decay rate
    ema_decay = 0.9999

    for epoch in range(num_epochs):
        total_loss = 0
        unet.train()
        text_encoder.train()

        for batch_idx, (images, tokens) in enumerate(tqdm(train_dataloader, desc=f"Diffusion Epoch {epoch+1}/{num_epochs}")):
            images = images.to(device)
            tokens["input_ids"] = tokens["input_ids"].to(device)
            tokens["attention_mask"] = tokens["attention_mask"].to(device)

            # Get text embeddings
            text_embeddings = text_encoder(tokens["input_ids"], tokens["attention_mask"])

            # Get latent representations
            with torch.no_grad():
                latents, _, _ = vae.encode(images)

            # Sample random timesteps with improved sampling strategy
            batch_size = latents.shape[0]

            # Improved timestep sampling for better training
            if epoch < num_epochs // 3:
                # Early training: uniform sampling
                timesteps = torch.randint(0, T, (batch_size,), device=device).long()
            else:
                # Later training: more emphasis on harder timesteps
                weights = torch.linspace(0.8, 0.2, T, device=device)
                timesteps_float = torch.multinomial(weights, batch_size, replacement=True).float()
                timesteps = timesteps_float.long()

            # Add noise to latents
            noisy_latents, noise = forward_diffusion_sample(latents, timesteps, device)

            # Conditioning dropout for classifier-free guidance training
            guidance_dropout_prob = 0.1
            use_guidance = torch.rand(batch_size, device=device) > guidance_dropout_prob
            context = text_embeddings * use_guidance.float().unsqueeze(1)

            # Predict noise
            noise_pred = unet(noisy_latents, timesteps, context)

            # Combined L1 and L2 loss for better detail preservation
            loss_mse = F.mse_loss(noise_pred, noise)
            loss_l1 = F.l1_loss(noise_pred, noise)
            loss = 0.8 * loss_mse + 0.2 * loss_l1

            # Scale for gradient accumulation
            loss = loss / gradient_accumulation_steps

            # Backpropagation with gradient accumulation
            loss.backward()

            if (batch_idx + 1) % gradient_accumulation_steps == 0:
                # Gradient clipping
                torch.nn.utils.clip_grad_norm_(text_encoder.parameters(), 1.0)
                torch.nn.utils.clip_grad_norm_(unet.parameters(), 1.0)

                text_encoder_optimizer.step()
                unet_optimizer.step()

                text_encoder_optimizer.zero_grad()
                unet_optimizer.zero_grad()

                # Update EMA model
                with torch.no_grad():
                    for param, ema_param in zip(unet.parameters(), ema_unet.parameters()):
                        ema_param.data = ema_decay * ema_param.data + (1 - ema_decay) * param.data

            total_loss += loss.item() * gradient_accumulation_steps

        # Update learning rate schedulers
        text_scheduler.step()
        unet_scheduler.step()

        avg_loss = total_loss / len(train_dataloader)
        print(f"Diffusion Epoch {epoch+1}/{num_epochs}, Loss: {avg_loss:.4f}")

        # Save best model
        if avg_loss < best_loss:
            best_loss = avg_loss
            torch.save(text_encoder.state_dict(), "butterfly_text_encoder_best.pt")
            torch.save(unet.state_dict(), "butterfly_unet_best.pt")
            torch.save(ema_unet.state_dict(), "butterfly_ema_unet_best.pt")
            print(f"New best model saved with loss: {best_loss:.4f}")

        # Save checkpoint periodically
        if (epoch + 1) % 5 == 0 or epoch == num_epochs - 1:
            torch.save(text_encoder.state_dict(), f"butterfly_text_encoder_epoch_{epoch+1}.pt")
            torch.save(unet.state_dict(), f"butterfly_unet_epoch_{epoch+1}.pt")
            torch.save(ema_unet.state_dict(), f"butterfly_ema_unet_epoch_{epoch+1}.pt")

            # Generate a sample image to show progress
            sample_prompt = "a beautiful monarch butterfly"
            sample_image = generate_image(sample_prompt, model_type="ema", steps=50)
            sample_image.save(f"sample_butterfly_epoch_{epoch+1}.png")

    # Save final models
    torch.save(text_encoder.state_dict(), "butterfly_text_encoder.pt")
    torch.save(unet.state_dict(), "butterfly_diffusion_model.pt")
    torch.save(ema_unet.state_dict(), "butterfly_ema_diffusion_model.pt")

    # Load the best models for inference
    text_encoder.load_state_dict(torch.load("butterfly_text_encoder_best.pt"))
    unet.load_state_dict(torch.load("butterfly_unet_best.pt"))
    ema_unet.load_state_dict(torch.load("butterfly_ema_unet_best.pt"))
    print("Loaded best models for inference")

# Improved image generation function
@torch.no_grad()
def generate_image(prompt, model_type="ema", steps=100, guidance_scale=7.5, seed=None):
    # Set seed for reproducibility if provided
    if seed is not None:
        torch.manual_seed(seed)
        np.random.seed(seed)

    vae.eval()
    text_encoder.eval()

    # Select which UNet to use (regular or EMA)
    if model_type == "ema" and 'ema_unet' in globals():
        model = ema_unet
    else:
        model = unet
    model.eval()

    # Enhance the prompt for better results
    enhanced_prompt = f"a detailed photograph of {prompt} butterfly with vibrant colors, high resolution, professional lighting"

    tokens = tokenizer(
        enhanced_prompt,
        padding="max_length",
        max_length=77,
        truncation=True,
        return_tensors="pt"
    ).to(device)

    # Get text embeddings
    text_embeddings = text_encoder(tokens.input_ids, tokens.attention_mask)

    # For classifier-free guidance, we need to do two forward passes:
    # one with the conditioned input (text_embeddings), and one with unconditional input
    uncond_tokens = tokenizer(
        "",  # empty string is unconditional
        padding="max_length",
        max_length=77,
        truncation=True,
        return_tensors="pt"
    ).to(device)

    uncond_embeddings = text_encoder(uncond_tokens.input_ids, uncond_tokens.attention_mask)

    # Concatenate the unconditional and conditional embeddings
    text_embeddings = torch.cat([uncond_embeddings, text_embeddings])

    # Initialize latents
    latents = torch.randn(
        (1, 4, 16, 16),  # Match the latent size from VAE
        device=device
    )

    # Set up timesteps for DDIM sampling (faster and often better quality)
    timesteps = torch.linspace(T-1, 0, steps).long().to(device)

    # Denoise the latents
    for i, t in enumerate(tqdm(timesteps)):
        # Expand the latents for classifier-free guidance
        latent_model_input = torch.cat([latents] * 2)

        # Get previous timestep
        prev_t = timesteps[i+1] if i < len(timesteps) - 1 else torch.tensor(0).to(device)

        # Predict noise
        noise_pred = model(latent_model_input, torch.cat([t.unsqueeze(0)] * 2), text_embeddings)

        # Perform classifier-free guidance
        noise_pred_uncond, noise_pred_text = noise_pred.chunk(2)
        noise_pred = noise_pred_uncond + guidance_scale * (noise_pred_text - noise_pred_uncond)

        # Get alpha values
        alpha_t = alphas_cumprod[t]
        alpha_prev = alphas_cumprod[prev_t]

        # Predict the original sample
        pred_x0 = (latents - torch.sqrt(1 - alpha_t) * noise_pred) / torch.sqrt(alpha_t)

        # Get direction to next step
        direction = torch.sqrt(1 - alpha_prev) * noise_pred

        # Set new latents (DDIM step)
        latents = torch.sqrt(alpha_prev) * pred_x0 + direction

    # Decode the final latent
    with torch.no_grad():
        images = vae.decode(latents)

    # Process the images with slight gamma correction for better colors
    images = (images / 2 + 0.5).clamp(0, 1)  # [-1, 1] -> [0, 1]
    images = torch.pow(images, 0.9)  # Gamma correction for more vibrant colors

    images = images.cpu().permute(0, 2, 3, 1).numpy()
    images = (images * 255).astype(np.uint8)

    return Image.fromarray(images[0])

# Gradio interface with more options
def gradio_interface(prompt, guidance_scale=7.5, num_steps=50, seed=None):
    if seed is not None:
        try:
            seed = int(seed)
        except:
            seed = None

    # Default to "colorful butterfly" if no prompt
    if not prompt.strip():
        prompt = "colorful"

    # Add "butterfly" if not in prompt
    if "butterfly" not in prompt.lower():
        prompt = f"{prompt} butterfly"

    return generate_image(
        prompt,
        model_type="ema",
        steps=num_steps,
        guidance_scale=guidance_scale,
        seed=seed
    )

# Create Gradio interface
demo = gr.Interface(
    fn=gradio_interface,
    inputs=[
        gr.Textbox(label="Enter butterfly description (e.g., 'monarch' or 'blue morpho')"),
        gr.Slider(minimum=1.0, maximum=15.0, value=7.5, step=0.5, label="Guidance Scale (higher = more prompt adherence)"),
        gr.Slider(minimum=20, maximum=150, value=50, step=5, label="Sampling Steps (higher = more quality, slower)"),
        gr.Textbox(label="Seed for reproducibility (optional)", placeholder="Leave empty for random results")
    ],
    outputs=gr.Image(label="Generated butterfly"),
    title="Butterfly Generator",
    description="Generate butterfly images from text descriptions, fine-tuned on the Smithsonian Butterflies dataset.",
    examples=[
        ["monarch with orange and black patterns", 7.5, 50, 42],
        ["blue morpho with iridescent blue wings", 8.0, 70, 123],
        ["swallowtail butterfly with yellow and black wings", 9.0, 80, 456],
        ["red and black butterfly in natural habitat", 7.0, 50, 789]
    ]
)

# Run main program
if __name__ == "__main__":
    # First train VAE
    print("Training VAE...")
    train_vae(num_epochs=25)

    # Then train diffusion model
    print("Training diffusion model...")
    train_diffusion(num_epochs=30, gradient_accumulation_steps=2)

    # Launch the Gradio interface
    demo.launch()
```
